```python
import numpy as np
import math as math
```

```python
def compute_NSE(Q_obs, Q_sim):
    Q_obs_mean = Q_obs.mean()
    num = 0
    den = 0
    for i in range(Q_obs.size):
        num += ((Q_obs[i] - Q_sim[i])**2)
        den += ((Q_obs[i] - Q_obs_mean)**2)
    NSE = 1-num/den
    return NSE
```

```python
def compute_KGE(Q_obs, Q_sim):
    a = Q_sim.std()/Q_obs.std()
    b = Q_sim.mean()/Q_obs.mean()
    r = np.corrcoef(Q_obs, Q_sim)[0][1]
    KGE = 1-math.sqrt((r-1)**2+(a-1)**2+(b-1)**2)
    return [KGE,r,a,b]
```

```python
def compute_PBIAS(Q_obs, Q_sim):
    num = 0
    den = 0
    for i in range(Q_obs.size):
        num += (Q_obs[i] - Q_sim[i])
        den += Q_obs[i]
    PBIAS = 100*num/den
    return PBIAS
```
