```python
import pandas as pd
```

```python
sf_mmd_path = r"D:\Claude\flood_hydrology_modeling\data\03_streamflow\streamflow_mmd.csv"
prec_AGCD_path = r"D:\Claude\flood_hydrology_modeling\data\05_hydrometeorology\01_precipitation_timeseries\precipitation_AGCD.csv"
prec_SILO_path = r"D:\Claude\flood_hydrology_modeling\data\05_hydrometeorology\01_precipitation_timeseries\precipitation_SILO.csv"
morton_PET_path = r"D:\Claude\flood_hydrology_modeling\data\05_hydrometeorology\02_EvaporativeDemand_timeseries\et_morton_actual_SILO.csv"
out_path = r"D:\Claude\flood_hydrology_modeling\data\processed\prec_PET_sf.csv"
start_end = ["1970-05-28", "2021-06-30"]
```

```python
def process_csv_from_path(path, start_end):
    global start, end
    file = path.split("\\")
    print(f"reading {file[-1:][0]}")
    print("-------------------------------------------------------------")
    df = pd.read_csv(path)
    df['date'] = pd.to_datetime(df[['year', 'month', 'day']])
    col = df.columns.to_list()
    [col.remove(item) for item in ['114001A', 'date']]
    df = df.drop(columns=col)
    [start_index, end_index] = [df.index[df['date']==start_end[0]].to_list()[0], df.index[df['date']==start_end[1]].to_list()[0]]
    print("Total days of data: ",end_index-start_index+1)
    print("\n")
    df = df.iloc[start_index:end_index+1] 
    df = df.reset_index(drop=True) 
    return df  
```

```python
# prec_AGCD = process_csv_from_path(prec_AGCD_path)
sf_mmd = process_csv_from_path(sf_mmd_path, start_end)
prec_AGCD = process_csv_from_path(prec_AGCD_path, start_end)
prec_SILO = process_csv_from_path(prec_SILO_path, start_end)
PET_morton = process_csv_from_path(morton_PET_path, start_end)
# print(sf_mmd.head())
# print(prec_AGCD.head())
# print(PET_morton.head())
df = pd.DataFrame({'date': sf_mmd['date'], 'prec_AGCD':prec_AGCD['114001A'], 'prec_SILO':prec_SILO['114001A'],'PET_morton':PET_morton['114001A'], 'sf_mmd': sf_mmd['114001A']})
df['sf_mmd'] = df['sf_mmd'].where(df['sf_mmd'] > -99.99 + 1e-6)
df.to_csv(out_path, index=False)
print("processed csv written to: ", out_path)
# print(df)
```
